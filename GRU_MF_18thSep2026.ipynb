The code trains and evaluates two separate datasets:

Dataset A — with market impact

Training:

2015 - 2021

Testing:

2022 - 2025

Dataset B — without market impact

Training:

2015 - 2019

Testing:

2022 - 2025

The years 2020 - 2021 are therefore excluded from training in Dataset B.

GRU prediction is performed individually for each fund.
The target is NAV, not return.
Performance metrics: MAE, MSE, RMSE, MAPE and R².

```python
# ================================================================
# GRU MODEL FOR MUTUAL FUND NAV PREDICTION
# 5 Mutual Funds | Single CSV File
#
# Dataset A:
#   Training : 2005-2011
#   Testing  : 2012-2015
#
# Dataset B:
#   Training : 2005-2009
#   Testing  : 2012-2015
#
# Target: NAV
# Models: One GRU model per mutual fund
# Metrics: MAE, MSE, RMSE, MAPE, R2
# ================================================================

# ------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    mean_absolute_percentage_error,
    r2_score
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping


# ------------------------------------------------
# 2. LOAD DATA
# ------------------------------------------------

FILE_PATH = "/content/drive/MyDrive/Datasets/GRU/cleaned_mutual_fund_navs_20260127_095249.csv"

df = pd.read_csv(FILE_PATH)

# Convert Date column to datetime
df["date"] = pd.to_datetime(df["date"])

# Sort chronologically
df = df.sort_values("date").reset_index(drop=True)

print("Dataset shape:", df.shape)
print(df.head())
print(df.tail())


# ------------------------------------------------
# 3. DEFINE THE FIVE MUTUAL FUNDS
# ------------------------------------------------

fund_columns = [
    'SBI Large Cap Fund - Direct Plan', 'HDFC Flexi Cap Fund - Growth Plan',
       'Axis Midcap Fund - Direct Plan',
       'ICICI Prudential Infrastructure Fund - Growth',
       'quant Small Cap Fund - IDCW Option - Regular Plan'
]

print("\nFunds used:")
for fund in fund_columns:
    print("-", fund)


# ------------------------------------------------
# 4. DEFINE DATASET PERIODS
# ------------------------------------------------

# Dataset A: WITH MARKET IMPACT
TRAIN_A_START = "2015-01-01"
TRAIN_A_END   = "2021-12-31"



# Dataset B: WITHOUT MARKET IMPACT
TRAIN_B_START = "2015-01-01"
TRAIN_B_END   = "2019-12-31"



# SAME TEST PERIOD FOR BOTH
TEST_START = "2022-01-01"
TEST_END   = "2025-05-31"


# ------------------------------------------------
# 5. GRU PARAMETERS
# ------------------------------------------------

LOOKBACK = 100

EPOCHS = 100
BATCH_SIZE = 32

# Random seed for reproducibility
np.random.seed(42)


# ------------------------------------------------
# 6. FUNCTION TO CREATE SEQUENCES
# ------------------------------------------------

def create_sequences(data, lookback):

    X = []
    y = []

    for i in range(lookback, len(data)):

        X.append(data[i - lookback:i])

        y.append(data[i])

    return np.array(X), np.array(y)


# ------------------------------------------------
# 7. FUNCTION TO BUILD GRU MODEL
# ------------------------------------------------

def build_gru_model():

    model = Sequential()

    model.add(
        GRU(
            64,
            return_sequences=True,
            input_shape=(LOOKBACK, 1)
        )
    )

    model.add(Dropout(0.2))

    model.add(
        GRU(
            32,
            return_sequences=False
        )
    )

    model.add(Dropout(0.2))

    model.add(Dense(16, activation="relu"))

    model.add(Dense(1))

    model.compile(
        optimizer="adam",
        loss="mean_squared_error"
    )

    return model


# ------------------------------------------------
# 8. FUNCTION FOR TRAINING AND PREDICTION
# ------------------------------------------------

def run_gru_for_fund(
    df,
    fund,
    train_start,
    train_end,
    test_start,
    test_end
):

    print("\n" + "=" * 70)
    print("Fund:", fund)
    print("Training:", train_start, "to", train_end)
    print("Testing :", test_start, "to", test_end)
    print("=" * 70)

    # ------------------------------------------------
    # Extract training and testing data
    # ------------------------------------------------

    train_df = df[
        (df["date"] >= train_start) &
        (df["date"] <= train_end)
    ][["date", fund]].dropna()

    test_df = df[
        (df["date"] >= test_start) &
        (df["date"] <= test_end)
    ][["date", fund]].dropna()

    print("Training observations:", len(train_df))
    print("Testing observations :", len(test_df))

    # ------------------------------------------------
    # Extract NAV values
    # ------------------------------------------------

    train_values = train_df[fund].values.reshape(-1, 1)

    test_values = test_df[fund].values.reshape(-1, 1)

    # ------------------------------------------------
    # Scale NAV values
    # IMPORTANT:
    # Fit scaler ONLY on training data
    # ------------------------------------------------

    scaler = MinMaxScaler()

    train_scaled = scaler.fit_transform(train_values)

    # For test data, use the same scaler
    test_scaled = scaler.transform(test_values)

    # ------------------------------------------------
    # Create training sequences
    # ------------------------------------------------

    X_train, y_train = create_sequences(
        train_scaled,
        LOOKBACK
    )

    # ------------------------------------------------
    # Create test sequences
    #
    # To predict the first test observation, we need
    # the last LOOKBACK observations from training.
    # ------------------------------------------------

    combined = np.concatenate(
        (train_scaled[-LOOKBACK:], test_scaled),
        axis=0
    )

    X_test, y_test = create_sequences(
        combined,
        LOOKBACK
    )

    print("X_train shape:", X_train.shape)
    print("y_train shape:", y_train.shape)

    print("X_test shape :", X_test.shape)
    print("y_test shape :", y_test.shape)

    # ------------------------------------------------
    # Build GRU model
    # ------------------------------------------------

    model = build_gru_model()

    model.summary()

    # ------------------------------------------------
    # Early stopping
    # ------------------------------------------------

    early_stop = EarlyStopping(
        monitor="val_loss",
        patience=10,
        restore_best_weights=True
    )

    # ------------------------------------------------
    # Train model
    # ------------------------------------------------

    history = model.fit(
        X_train,
        y_train,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        validation_split=0.1,
        callbacks=[early_stop],
        shuffle=False,
        verbose=1
    )

    # ------------------------------------------------
    # Predict test NAV
    # ------------------------------------------------

    predictions_scaled = model.predict(
        X_test,
        verbose=0
    )

    # ------------------------------------------------
    # Convert predictions back to original NAV scale
    # ------------------------------------------------

    predictions = scaler.inverse_transform(
        predictions_scaled
    )

    actual = scaler.inverse_transform(
        y_test.reshape(-1, 1)
    )

    # ------------------------------------------------
    # Performance metrics
    # ------------------------------------------------

    mae = mean_absolute_error(
        actual,
        predictions
    )

    mse = mean_squared_error(
        actual,
        predictions
    )

    rmse = np.sqrt(mse)

    mape = mean_absolute_percentage_error(
        actual,
        predictions
    ) * 100

    r2 = r2_score(
        actual,
        predictions
    )

    print("\nPerformance Metrics")
    print("----------------------------")
    print("MAE  :", round(mae, 4))
    print("MSE  :", round(mse, 4))
    print("RMSE :", round(rmse, 4))
    print("MAPE :", round(mape, 4), "%")
    print("R²   :", round(r2, 4))

    # ------------------------------------------------
    # Create prediction dataframe
    # ------------------------------------------------

    # Because the first test sequence starts with the
    # final LOOKBACK observations from training, the
    # predictions correspond exactly to the test period.

    results = pd.DataFrame({
        "date": test_df["date"].values,
        "Actual_NAV": actual.flatten(),
        "Predicted_NAV": predictions.flatten()
    })

    # ------------------------------------------------
    # Plot Actual vs Predicted NAV
    # ------------------------------------------------

    plt.figure(figsize=(12, 5))

    plt.plot(
        results["date"],
        results["Actual_NAV"],
        label="Actual NAV"
    )

    plt.plot(
        results["date"],
        results["Predicted_NAV"],
        label="Predicted NAV"
    )

    plt.title(
        f"GRU NAV Prediction - {fund}"
    )

    plt.xlabel("Date")
    plt.ylabel("NAV")

    plt.legend()

    plt.grid(True)

    plt.tight_layout()

    plt.show()

    # ------------------------------------------------
    # Plot training and validation loss
    # ------------------------------------------------

    plt.figure(figsize=(10, 5))

    plt.plot(
        history.history["loss"],
        label="Training Loss"
    )

    plt.plot(
        history.history["val_loss"],
        label="Validation Loss"
    )

    plt.title(
        f"GRU Training Loss - {fund}"
    )

    plt.xlabel("Epoch")

    plt.ylabel("MSE Loss")

    plt.legend()

    plt.grid(True)

    plt.tight_layout()

    plt.show()

    return {
        "model": model,
        "scaler": scaler,
        "results": results,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "MAPE": mape,
        "R2": r2
    }


# ================================================================
# 9. DATASET A
#    WITH MARKET IMPACT
#    TRAIN: 2005-2011
#    TEST : 2012-2015
# ================================================================

print("\n\n")
print("#" * 80)
print("DATASET A - WITH MARKET IMPACT")
print("#" * 80)

results_A = {}

for fund in fund_columns:

    results_A[fund] = run_gru_for_fund(
        df=df,
        fund=fund,
        train_start=TRAIN_A_START,
        train_end=TRAIN_A_END,
        test_start=TEST_START,
        test_end=TEST_END
    )


# ================================================================
# 10. DATASET B
#     WITHOUT MARKET IMPACT
#     TRAIN: 2005-2009
#     TEST : 2012-2015
# ================================================================

print("\n\n")
print("#" * 80)
print("DATASET B - WITHOUT MARKET IMPACT")
print("#" * 80)

results_B = {}

for fund in fund_columns:

    results_B[fund] = run_gru_for_fund(
        df=df,
        fund=fund,
        train_start=TRAIN_B_START,
        train_end=TRAIN_B_END,
        test_start=TEST_START,
        test_end=TEST_END
    )


# ================================================================
# 11. CREATE PERFORMANCE TABLE - DATASET A
# ================================================================

performance_A = []

for fund in fund_columns:

    performance_A.append({
        "Fund": fund,
        "MAE": results_A[fund]["MAE"],
        "MSE": results_A[fund]["MSE"],
        "RMSE": results_A[fund]["RMSE"],
        "MAPE (%)": results_A[fund]["MAPE"],
        "R2": results_A[fund]["R2"]
    })

performance_A = pd.DataFrame(performance_A)

print("\n")
print("=" * 80)
print("GRU PERFORMANCE - DATASET A")
print("=" * 80)

print(
    performance_A.round(4).to_string(index=False)
)


# ================================================================
# 12. CREATE PERFORMANCE TABLE - DATASET B
# ================================================================

performance_B = []

for fund in fund_columns:

    performance_B.append({
        "Fund": fund,
        "MAE": results_B[fund]["MAE"],
        "MSE": results_B[fund]["MSE"],
        "RMSE": results_B[fund]["RMSE"],
        "MAPE (%)": results_B[fund]["MAPE"],
        "R2": results_B[fund]["R2"]
    })

performance_B = pd.DataFrame(performance_B)

print("\n")
print("=" * 80)
print("GRU PERFORMANCE - DATASET B")
print("=" * 80)

print(
    performance_B.round(4).to_string(index=False)
)


# ================================================================
# 13. COMPARISON OF DATASET A AND DATASET B
# ================================================================

comparison = []

for fund in fund_columns:

    comparison.append({

        "Fund": fund,

        "A_RMSE": results_A[fund]["RMSE"],
        "B_RMSE": results_B[fund]["RMSE"],

        "A_MAE": results_A[fund]["MAE"],
        "B_MAE": results_B[fund]["MAE"],

        "A_MAPE (%)": results_A[fund]["MAPE"],
        "B_MAPE (%)": results_B[fund]["MAPE"],

        "A_R2": results_A[fund]["R2"],
        "B_R2": results_B[fund]["R2"]

    })

comparison = pd.DataFrame(comparison)

print("\n")
print("=" * 80)
print("DATASET A vs DATASET B - GRU")
print("=" * 80)

print(
    comparison.round(4).to_string(index=False)
)


# ================================================================
# 14. SAVE PERFORMANCE RESULTS
# ================================================================

performance_A.to_csv(
    "GRU_Performance_Dataset_A.csv",
    index=False
)

performance_B.to_csv(
    "GRU_Performance_Dataset_B.csv",
    index=False
)

comparison.to_csv(
    "GRU_A_vs_B_Comparison.csv",
    index=False
)


# ================================================================
# 15. SAVE PREDICTIONS FOR ALL FIVE FUNDS
# ================================================================

for fund in fund_columns:

    safe_name = fund.replace(" ", "_")

    results_A[fund]["results"].to_csv(
        f"GRU_A_{safe_name}_Predictions.csv",
        index=False
    )

    results_B[fund]["results"].to_csv(
        f"GRU_B_{safe_name}_Predictions.csv",
        index=False
    )

print("\nAll GRU results and predictions have been saved.")
```

### **GRU Model Setup & Hyperparameters Summary**

#### **1. Data Splitting Strategy**
Two distinct experimental configurations are evaluated to analyze the effect of excluding specific periods (such as market anomalies or high-impact events):

| Dataset | Market Impact Treatment | Training Period | Testing Period | Excluded Period |
| :--- | :--- | :--- | :--- | :--- |
| **Dataset A** | **With** Market Impact | 2015-01-01 to 2021-12-31 | 2022-01-01 to 2025-05-31 | None |
| **Dataset B** | **Without** Market Impact | 2015-01-01 to 2019-12-31 | 2022-01-01 to 2025-05-31 | 2020 to 2021 |

---

#### **2. Model Architecture & Hyperparameters**
Individual GRU models are designed and trained for each of the **5 Mutual Funds** using a **sequential, many-to-one** forecasting scheme.

* **Target Variable**: Net Asset Value (NAV)
* **Lookback Window**: `100` days (uses the past 100 observations to predict the next day's NAV)
* **Scaling**: `MinMaxScaler(feature_range=(0, 1))` fit exclusively on the corresponding training dataset
* **Model Layers**:
  1. **GRU Layer 1**: 64 units (returns sequences)
  2. **Dropout Layer 1**: 20% rate
  3. **GRU Layer 2**: 32 units (returns final state only)
  4. **Dropout Layer 2**: 20% rate
  5. **Dense (Hidden) Layer**: 16 units with ReLU activation
  6. **Dense (Output) Layer**: 1 unit (Linear activation for regression)
* **Optimizer**: Adam
* **Loss Function**: Mean Squared Error (MSE)

---

#### **3. Training & Regularization Settings**
* **Epochs**: Maximum of `100`
* **Batch Size**: `32`
* **Validation Split**: `10%` of training data reserved for validation (sequential validation set)
* **Shuffle**: `False` (disabled to preserve the sequence ordering in time-series data)
* **Early Stopping**: Monitored on validation loss (`val_loss`) with a patience of `10` epochs, restoring the best model weights.
* **Reproducibility Seed**: `np.random.seed(42)`

---

#### **4. Evaluation Metrics**
The models are evaluated on the test set across five performance dimensions:
1. **MAE** (Mean Absolute Error)
2. **MSE** (Mean Squared Error)
3. **RMSE** (Root Mean Squared Error)
4. **MAPE** (Mean Absolute Percentage Error)
5. **R²** (Coefficient of Determination)
