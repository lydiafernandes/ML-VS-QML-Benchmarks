```python
# ================================================================
# LSTM FOR MUTUAL FUND NAV PREDICTION
# Five Mutual Funds - Individual Models
#
# Dataset A: With Market Impact
# Training : 2015-2021
# Testing  : 2022-2025
#
# Dataset B: Without Market Impact
# Training : 2015-2019
# Testing  : 2022-2025
#
# Target: NAV
# ================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping


# ================================================================
# 1. LOAD DATA
# ================================================================

FILE_PATH = "/content/drive/MyDrive/Datasets/LSTM_latest/cleaned_mutual_fund_navs_20260127_095249.csv"

df = pd.read_csv(FILE_PATH)

# Convert Date column
df["date"] = pd.to_datetime(df["date"])

# Sort chronologically
df = df.sort_values("date").reset_index(drop=True)

print("\nDataset Information")
print("-------------------")
print(df.head())
print(df.tail())

print("\nDate range:")
print(df["date"].min(), "to", df["date"].max())


# ================================================================
# 2. IDENTIFY FIVE MUTUAL FUNDS
# ================================================================

fund_columns = [
    'SBI Large Cap Fund - Direct Plan', 'HDFC Flexi Cap Fund - Growth Plan',
       'Axis Midcap Fund - Direct Plan',
       'ICICI Prudential Infrastructure Fund - Growth',
       'quant Small Cap Fund - IDCW Option - Regular Plan'
]

print("\nMutual Funds Used:")
for fund in fund_columns:
    print(fund)


# ================================================================
# 3. DEFINE EXPERIMENTAL PERIODS
# ================================================================

# Dataset A
TRAIN_A_START = "2015-01-01"
TRAIN_A_END   = "2021-12-31"

# Dataset B
TRAIN_B_START = "2015-01-01"
TRAIN_B_END   = "2019-12-31"

# SAME TESTING PERIOD FOR BOTH
TEST_START = "2022-01-01"
TEST_END   = "2025-05-31"


# ================================================================
# 4. LSTM PARAMETERS
# ================================================================

LOOKBACK = 100

EPOCHS = 100
BATCH_SIZE = 32

LEARNING_RATE = 0.001

RANDOM_SEED = 42

np.random.seed(RANDOM_SEED)


# ================================================================
# 5. CREATE LSTM SEQUENCES
# ================================================================

def create_sequences(data, lookback):

    X = []
    y = []

    for i in range(lookback, len(data)):

        X.append(data[i-lookback:i])
        y.append(data[i])

    return np.array(X), np.array(y)


# ================================================================
# 6. BUILD LSTM MODEL
# ================================================================

def build_lstm_model():

    model = Sequential()

    model.add(
        LSTM(
            50,
            activation="relu",
            return_sequences=True,
            input_shape=(LOOKBACK, 1)
        )
    )

    model.add(Dropout(0.20))

    model.add(
        LSTM(
            60,
            activation="relu",
            return_sequences=True
        )
    )

    model.add(Dropout(0.30))

    model.add(
        LSTM(
            80,
            activation="relu",
            return_sequences=True
        )
    )

    model.add(Dropout(0.40))

    model.add(
        LSTM(
            120,
            activation="relu"
        )
    )

    model.add(Dropout(0.50))

    model.add(Dense(1))

    optimizer = Adam(
        learning_rate=LEARNING_RATE
    )

    model.compile(
        optimizer=optimizer,
        loss="mse"
    )

    return model


# ================================================================
# 7. FUNCTION TO CALCULATE METRICS
# ================================================================

def calculate_metrics(actual, predicted):

    mae = mean_absolute_error(
        actual,
        predicted
    )

    mse = mean_squared_error(
        actual,
        predicted
    )

    rmse = np.sqrt(mse)

    # Avoid division by zero
    mask = actual != 0

    mape = np.mean(
        np.abs(
            (actual[mask] - predicted[mask])
            / actual[mask]
        )
    ) * 100

    r2 = r2_score(
        actual,
        predicted
    )

    return mae, mse, rmse, mape, r2


# ================================================================
# 8. TRAIN AND TEST ONE FUND
# ================================================================

def run_lstm_for_fund(
    df,
    fund,
    train_start,
    train_end,
    test_start,
    test_end,
    dataset_name
):

    print("\n" + "=" * 70)
    print(f"{dataset_name} - {fund}")
    print("=" * 70)


    # ------------------------------------------------------------
    # Extract required dates
    # ------------------------------------------------------------

    train_df = df[
        (df["date"] >= train_start) &
        (df["date"] <= train_end)
    ][["date", fund]].copy()

    test_df = df[
        (df["date"] >= test_start) &
        (df["date"] <= test_end)
    ][["date", fund]].copy()


    # Remove missing NAV values
    train_df = train_df.dropna()
    test_df = test_df.dropna()


    print(
        f"Training observations: {len(train_df)}"
    )

    print(
        f"Testing observations : {len(test_df)}"
    )


    # ------------------------------------------------------------
    # Extract NAV
    # ------------------------------------------------------------

    train_values = train_df[fund].values.reshape(-1, 1)

    test_values = test_df[fund].values.reshape(-1, 1)


    # ------------------------------------------------------------
    # IMPORTANT:
    # Fit scaler ONLY on training data
    # ------------------------------------------------------------

    scaler = MinMaxScaler()

    train_scaled = scaler.fit_transform(
        train_values
    )

    test_scaled = scaler.transform(
        test_values
    )


    # ------------------------------------------------------------
    # Create training sequences
    # ------------------------------------------------------------

    X_train, y_train = create_sequences(
        train_scaled,
        LOOKBACK
    )


    # ------------------------------------------------------------
    # Create testing sequences
    #
    # To predict the first test observation, we need the
    # previous LOOKBACK observations.
    #
    # Therefore we use the last LOOKBACK training observations
    # together with the test observations.
    # ------------------------------------------------------------

    combined = np.concatenate(
        [
            train_scaled[-LOOKBACK:],
            test_scaled
        ]
    )

    X_test, y_test = create_sequences(
        combined,
        LOOKBACK
    )


    print(
        "X_train shape:",
        X_train.shape
    )

    print(
        "X_test shape :",
        X_test.shape
    )


    # ------------------------------------------------------------
    # Build model
    # ------------------------------------------------------------

    model = build_lstm_model()


    # ------------------------------------------------------------
    # Early stopping
    # ------------------------------------------------------------

    early_stopping = EarlyStopping(
        monitor="val_loss",
        patience=15,
        restore_best_weights=True
    )


    # ------------------------------------------------------------
    # Train model
    # ------------------------------------------------------------

    history = model.fit(
        X_train,
        y_train,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        validation_split=0.10,
        shuffle=False,
        callbacks=[early_stopping],
        verbose=1
    )


    # ------------------------------------------------------------
    # Predict
    # ------------------------------------------------------------

    predictions_scaled = model.predict(
        X_test,
        verbose=0
    )


    # ------------------------------------------------------------
    # Convert predictions back to original NAV scale
    # ------------------------------------------------------------

    predictions = scaler.inverse_transform(
        predictions_scaled
    ).flatten()

    actual = scaler.inverse_transform(
        y_test.reshape(-1, 1)
    ).flatten()


    # ------------------------------------------------------------
    # Calculate performance metrics
    # ------------------------------------------------------------

    mae, mse, rmse, mape, r2 = calculate_metrics(
        actual,
        predictions
    )


    print("\nPerformance Metrics")
    print("-------------------")

    print(f"MAE  : {mae:.4f}")
    print(f"MSE  : {mse:.4f}")
    print(f"RMSE : {rmse:.4f}")
    print(f"MAPE : {mape:.4f}%")
    print(f"R²   : {r2:.4f}")


    # ------------------------------------------------------------
    # Prediction table
    # ------------------------------------------------------------

    result_df = pd.DataFrame({

        "date": test_df["date"].values,

        "Actual_NAV": actual,

        "Predicted_NAV": predictions

    })


    # ------------------------------------------------------------
    # Plot Actual vs Predicted NAV
    # ------------------------------------------------------------

    plt.figure(figsize=(12, 6))

    plt.plot(
        result_df["date"],
        result_df["Actual_NAV"],
        label="Actual NAV"
    )

    plt.plot(
        result_df["date"],
        result_df["Predicted_NAV"],
        label="Predicted NAV"
    )

    plt.title(
        f"LSTM NAV Prediction - {fund} - {dataset_name}"
    )

    plt.xlabel("Date")
    plt.ylabel("NAV")

    plt.legend()

    plt.grid(True)

    plt.tight_layout()

    plt.show()


    # ------------------------------------------------------------
    # Plot training loss
    # ------------------------------------------------------------

    plt.figure(figsize=(10, 5))

    plt.plot(
        history.history["loss"],
        label="Training Loss"
    )

    plt.plot(
        history.history["val_loss"],
        label="Validation Loss"
    )

    plt.title(
        f"LSTM Training Loss - {fund} - {dataset_name}"
    )

    plt.xlabel("Epoch")

    plt.ylabel("MSE Loss")

    plt.legend()

    plt.grid(True)

    plt.tight_layout()

    plt.show()


    return {

        "Fund": fund,

        "Dataset": dataset_name,

        "MAE": mae,

        "MSE": mse,

        "RMSE": rmse,

        "MAPE": mape,

        "R2": r2,

        "Actual": actual,

        "Predicted": predictions,

        "Dates": result_df["date"],

        "Model": model,

        "Scaler": scaler,

        "History": history

    }


# ================================================================
# 9. RUN DATASET A
# ================================================================

results_A = []

for fund in fund_columns:

    result = run_lstm_for_fund(

        df=df,

        fund=fund,

        train_start=TRAIN_A_START,

        train_end=TRAIN_A_END,

        test_start=TEST_START,

        test_end=TEST_END,

        dataset_name="Dataset A - With Market Impact"

    )

    results_A.append(result)


# ================================================================
# 10. RUN DATASET B
# ================================================================

results_B = []

for fund in fund_columns:

    result = run_lstm_for_fund(

        df=df,

        fund=fund,

        train_start=TRAIN_B_START,

        train_end=TRAIN_B_END,

        test_start=TEST_START,

        test_end=TEST_END,

        dataset_name="Dataset B - Without Market Impact"

    )

    results_B.append(result)


# ================================================================
# 11. CREATE PERFORMANCE TABLE - DATASET A
# ================================================================

metrics_A = pd.DataFrame({

    "Fund": [
        r["Fund"] for r in results_A
    ],

    "MAE": [
        r["MAE"] for r in results_A
    ],

    "MSE": [
        r["MSE"] for r in results_A
    ],

    "RMSE": [
        r["RMSE"] for r in results_A
    ],

    "MAPE": [
        r["MAPE"] for r in results_A
    ],

    "R2": [
        r["R2"] for r in results_A
    ]

})


metrics_A["Dataset"] = "With Market Impact"


# ================================================================
# 12. CREATE PERFORMANCE TABLE - DATASET B
# ================================================================

metrics_B = pd.DataFrame({

    "Fund": [
        r["Fund"] for r in results_B
    ],

    "MAE": [
        r["MAE"] for r in results_B
    ],

    "MSE": [
        r["MSE"] for r in results_B
    ],

    "RMSE": [
        r["RMSE"] for r in results_B
    ],

    "MAPE": [
        r["MAPE"] for r in results_B
    ],

    "R2": [
        r["R2"] for r in results_B
    ]

})


metrics_B["Dataset"] = "Without Market Impact"


# ================================================================
# 13. COMBINE RESULTS
# ================================================================

final_metrics = pd.concat(
    [
        metrics_A,
        metrics_B
    ],
    ignore_index=True
)


# Reorder columns

final_metrics = final_metrics[
    [
        "Fund",
        "Dataset",
        "MAE",
        "MSE",
        "RMSE",
        "MAPE",
        "R2"
    ]
]


print("\n")
print("=" * 80)
print("FINAL LSTM PERFORMANCE COMPARISON")
print("=" * 80)

print(
    final_metrics.to_string(
        index=False
    )
)


# ================================================================
# 14. SAVE PERFORMANCE RESULTS
# ================================================================

final_metrics.to_csv(
    "LSTM_NAV_Performance_Comparison.csv",
    index=False
)


# ================================================================
# 15. SAVE ALL PREDICTIONS
# ================================================================

prediction_tables = []


# Dataset A

for r in results_A:

    temp = pd.DataFrame({

        "Date": r["Dates"],

        "Fund": r["Fund"],

        "Dataset": "With Market Impact",

        "Actual_NAV": r["Actual"],

        "Predicted_NAV": r["Predicted"]

    })

    prediction_tables.append(temp)


# Dataset B

for r in results_B:

    temp = pd.DataFrame({

        "Date": r["Dates"],

        "Fund": r["Fund"],

        "Dataset": "Without Market Impact",

        "Actual_NAV": r["Actual"],

        "Predicted_NAV": r["Predicted"]

    })

    prediction_tables.append(temp)


all_predictions = pd.concat(
    prediction_tables,
    ignore_index=True
)


all_predictions.to_csv(
    "LSTM_NAV_Predictions.csv",
    index=False
)


print("\nFiles generated:")
print("1. LSTM_NAV_Performance_Comparison.csv")
print("2. LSTM_NAV_Predictions.csv")
```
